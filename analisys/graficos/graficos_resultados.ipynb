# Gráficos dos resultados da EDA

Este notebook executa o script `graficos_resultados.py`, que lê os resultados em `analisys/resultados/` e salva os gráficos em `analisys/graficos/resultados/`.

```python
from pathlib import Path
import runpy

script = Path.cwd() / "graficos_resultados.py"
if not script.exists():
    script = Path.cwd() / "analisys" / "graficos" / "graficos_resultados.py"
if not script.exists():
    raise FileNotFoundError(f"Script não encontrado: {script}")

runpy.run_path(str(script), run_name="__main__")
```
